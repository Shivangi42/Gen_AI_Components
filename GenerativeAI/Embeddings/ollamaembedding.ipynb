Ollama

Ollama supports embedding models,making it possible to build retrieval augmented generation(RAG) applications that combine texts prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings=OllamaEmbeddings(model="gemma:2b") #by default it uses llama 2
```

```python
embeddings
```

```python
from langchain_community.embeddings import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="nomic-embed-text")

r1 = embeddings.embed_documents(
    [
        "Alpha is the first letter of Greek alphabet",
        "Beta is the second letter of Greek alphabet"
    ]
)

print(r1)
```

```python
print(r1[1])
```

```python
embeddings.embed_query("What is the second letter of Greek alphabet")
```

```python
embeddings=OllamaEmbeddings(model="mxbai-embed-large")
text="This is a test document"
query_result=embeddings.embed_query(text)
query_result
```
